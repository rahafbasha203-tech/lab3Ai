```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

file_path = '/content/hypertension_dataset.csv'
df = pd.read_csv(file_path)

print("تم تحميل البيانات بنجاح")
```

```python
X = df.iloc[:, :-1]
y = df.iloc[:, -1]

print("استكشاف البيانات")
print(f"عدد العينات (Samples): {df.shape[0]}")
print(f"عدد الميزات (Features): {X.shape[1]}")
print(f"الفئات المستهدفة (Classes): {np.unique(y)}")
print("\n أول 5 صفوف من مجموعة البيانات")
df.head()
```

```python
df_encoded = df.copy()

for col in df_encoded.columns:
    if df_encoded[col].dtype == 'object':
        df_encoded[col] = df_encoded[col].astype('category').cat.codes

X = df_encoded.iloc[:, :-1]
y = df_encoded.iloc[:, -1]

X_values = X.values
y_values = y.values

X_train, X_test, y_train, y_test = train_test_split(
    X_values, y_values, test_size=0.2, random_state=42
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("تم تحويل النصوص إلى أرقام وتقييس البيانات بنجاح")
```

```python
num_classes = len(np.unique(y_values))
input_dim = X_train_scaled.shape[1]

model = Sequential([
    Dense(64, activation='relu', input_shape=(input_dim,)),
    Dense(32, activation='relu'),
    Dense(1 if num_classes <= 2 else num_classes,
          activation='sigmoid' if num_classes <= 2 else 'softmax')
])

print("تم بناء هيكل النموذج بنجاح")
```

```python
if num_classes <= 2:
    loss_function = 'binary_crossentropy'
else:
    loss_function = 'sparse_categorical_crossentropy'

model.compile(
    optimizer='adam',
    loss=loss_function,
    metrics=['accuracy']
)

model.summary()
```

```python
history = model.fit(
    X_train_scaled,
    y_train,
    epochs=50,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)
```

```python
print("تقييم النموذج على بيانات الاختبار")
test_loss, test_accuracy = model.evaluate(X_test_scaled, y_test, verbose=0)

print(f"قيمة الخسارة (Test Loss): {test_loss:.4f}")
print(f"دقة الاختبار (Test Accuracy): {test_accuracy * 100:.2f}%")
```
